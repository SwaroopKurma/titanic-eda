```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns


df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

print(df.shape)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['AgeGroup'] = np.where(df['Age']<18,'Child','Adult')

print(df.shape)
print(df.head())
print(df)
```

```python
df['Survived'].value_counts().plot(kind='bar', color=['red','green'])
plt.title('People survived')
plt.xlabel('0=Died,1=Survived')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.hist(df['Age'],bins=30,color='steelblue',edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
df[df['Survived']==1]['Age'].hist(bins=30,alpha=0.6,color='green',label='Survived')
df[df['Survived']==0]['Age'].hist(bins=30,alpha=0.6,color='red',label='Died')

plt.title('people survival by Age')
plt.xlabel('Age')
plt.legend()
plt.show()
```

```python
sns.countplot(x='Pclass',hue='Survived', data=df, palette='Set2')
plt.title('Survival by Class')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='2f')
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.boxplot(x='Pclass',y='Age',data=df)
plt.title('Age Per Classes')
plt.show()
```

```python
df.groupby('Pclass')['Fare'].mean().plot(kind='bar')
plt.xticks(rotation=0)
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='pie')
plt.show()
```

```python
sns.scatterplot(x='Age',y='Fare',hue='Survived', data=df)
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
df['Survived'].value_counts().plot(kind='bar',color=['red','green'])
plt.title('Survived people')
plt.xticks(rotation=0)

plt.subplot(1,2,2)
sns.countplot(x='Pclass', hue='Survived', data=df , palette='Set2')
plt.title('Survival by Class')

plt.tight_layout()
plt.show()
```
